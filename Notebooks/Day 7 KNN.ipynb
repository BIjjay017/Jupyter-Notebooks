```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv("fraud_0.1origbase.csv")
```

```python
data.head()
```

```python
data.isFraud.value_counts()
```

```python
 data.columns
```

```python
data.drop({'step','nameOrig','nameDest'},inplace = True,axis=1)
```

```python
data
```

```python
data_non_fraud=data[data['isFraud'] ==0][:1000]
```

```python
data_non_fraud
```

```python
data_fraud = data[data['isFraud']==1]
```

```python
data_fraud
```

```python
#concat the above two tables
data = pd.concat([data_non_fraud,data_fraud])
```

```python
data
```

```python
label_enc = LabelEncoder()
```

```python
data['type']= label_enc.fit_transform(data['type'])
```

```python
x=data.drop('isFraud',axis=1)
y=data['isFraud']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.25,stratify =y ,random_state =2)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_class = KNeighborsClassifier(n_neighbors=7)
```

```python
help(knn_class)
```

```python
knn_class.fit(x_train,y_train)
```

```python
y_pred = knn_class.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
accuracy = accuracy_score(y_pred, y_test.values)
```

```python
accuracy
```

```python
class_report = classification_report(y_pred, y_test.values)
```

```python
print(class_report)
```

```python
con_mat = confusion_matrix(y_pred, y_test.values)
```
