# Association Analysis

```python
import numpy as np 
import pandas as pd 
from mlxtend.frequent_patterns import apriori, association_rules 
```

```python
cust_data = pd.read_csv('OnlineRetail.csv') 
```

```python
cust_data['Description'] = cust_data['Description'].str.strip()
```

```python
cust_data.dropna(axis = 0, subset =['InvoiceNo'], inplace = True) 
cust_data['InvoiceNo'] = cust_data['InvoiceNo'].astype('str') 
```

```python
cust_data['Country'].value_counts()
```

## Association analysis in Germany

```python
cust_data_Germany = (cust_data[cust_data['Country'] =="Germany"] 
		.groupby(['InvoiceNo', 'Description'])['Quantity'] 
		.sum().unstack().reset_index().fillna(0) 
		.set_index('InvoiceNo'))
```

```python
cust_data_Germany
```

### Encode the Description, i.e. items shopped together under same invoice

```python
def one_hot_encoding(x): 
    if(x<= 0): 
        return 0
    if(x>= 1): 
        return 1
```

```python
cust_data_Germany_encoded = cust_data_Germany.map(one_hot_encoding) 
```

```python
cust_data_Germany_encoded
```

```python
 frq_items = apriori(cust_data_Germany_encoded, min_support = 0.01, use_colnames = True) 

rules = association_rules(frq_items, metric ="lift", min_threshold = 1) 
rules = rules.sort_values(['confidence', 'lift'], ascending =[False, False]) 
```

```python
rules.head()
```

### Experiment with different min_support values

```python
for min_sup in [0.01, 0.02, 0.05]:
    frq_items = apriori(cust_data_Germany_encoded, min_support=min_sup, use_colnames=True)
    rules = association_rules(frq_items, metric="lift", min_threshold=1)
    print(f"Number of rules with min_support {min_sup}: {len(rules)}")
```
